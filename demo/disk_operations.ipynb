# Disk Operations

Just a brief demo to preempt any confusion around the various available libraries for disk operations.

```python
import os,glob
from pathlib import Path
```

## `os`

`os` is a built-in library with many basic commands for disk operations:

```python
os.getcwd()
```

```python
os.chdir('..')
os.getcwd()
```

```python
os.chdir('demo')
os.getcwd()
```

```python
os.listdir()
```

```python
sorted(os.listdir())
```

People use `os` because it has been around forever, so they're used to using it.

## `glob`

`glob` is just an alternative to `os`, but it supports wildcards:

```python
glob.glob('t*')
```

With `os`, you would have to filter the file names using other Python commands after obtaining the list. So people use `glob` for the convenience of built-in wildcards.

## `pathlib`

The big advantage of `pathlib` is that its path objects are platform-independent. Whereas `os` and `glob` take regular strings, with forward slashes or backslashes baked in dependent in your operating system architecture, `pathlib` allows you to create platform-agnostic objects that will work on any platform:

```python
xml = Path.cwd().parent / 'corpora' / 'echoe' / 'xml'
print(str(xml))
```

It is thanks to `pathlib` that I was able to draft all these demo notebooks in a way that runs on Windows as well as Linux and macOS.

Thus if you are writing code for your own use only, you can use whatever disk operations library you like. But if you ever share your code for use by others, or if you intend to use your code across platforms yourself, you are best off using `pathlib`. And really what that means is you probably want to use `pathlib` at all times so you will never have to think about which library to use. Because `pathlib` still supports wildcards:

```python
list(xml.glob('182*xml'))
```


One final note: you will occasionally run into libraries (e.g. `nltk`) that don't support `pathlib` objects. In those cases, simply convert the object to a string as in code cell 8 above.
