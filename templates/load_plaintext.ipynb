# Loading a Plaintext Corpus

This code template demonstrates how to load a folder of plaintext files into a corpus for further processing. The example assumes the folder structure created by running [`fetch_repo.ipynb`](https://github.com/langeslag/ehtc/blob/main/templates/fetch_repo.ipynb) from a folder of the same depth. For more documentation, see [the corresponding demo notebook](https://github.com/langeslag/ehtc/blob/main/demo/load_plaintext.ipynb).

```python
import os,glob
```

```python
# Point to a folder containing your corpus in plaintext format:
local = '../corpora/echoe/plaintext'
# Initialize an empty dictionary:
echoe = dict()
# Compile a list of files to be read in, and run the rest of the cell on a per-file basis:
for file in sorted(glob.glob(local + '/*.txt')):
    # Keep the base file name as an identifier:
    identifier = os.path.basename(file).replace('.txt', '')
    # Read in the document one line at a time:
    segments = open(file).read().splitlines()
    # Initialize an empty list of tokens:
    tokens = []
    # Run the next few lines on a per-line basis:
    for segment in segments:
        # If the line begins with an identifier (in this case set apart by a colon and a space), discard the identifier:
        if ': ' in segment:
            segment = segment.split(': ', 1)[1]
        # Tokenize the rest of the line and add the output to the list of tokens, and strip any spaces at line end:
        tokens.extend(segment.rstrip().split())
    # Finally, store the list of tokens in the dictionary, using the identifier as a key:
    echoe[identifier] = tokens
```

Having read in our tokens, we can now access them as follows (limiting to 20 for convenience):

```python
echoe['049B.11'][:20]
```
