# MediaWiki API

See the [demo notebook](https://github.com/langeslag/ehtc/blob/main/demo/mediawiki.ipynb) on this topic for guidance, and for an explanation of why we often need to use `BeautifulSoup` alongside `mediawiki` to access documents from wikis other than Wikpedia.

```python
from mediawiki import MediaWiki
from bs4 import BeautifulSoup
```

```python
wiki = MediaWiki(url='http://heimskringla.no/api.php')
#wiki = MediaWiki(url='https://secure.wikimedia.org/wikisource/en/w/api.php')
#wiki = MediaWiki(url='https://secure.wikimedia.org/wikisource/de/w/api.php')
```

```python
page = wiki.page('Völuspá')
html = page.html
soup = BeautifulSoup(html, 'html.parser')
# Inspection of the HTML source reveals the branch we want is heir to a single <dl> node:
poem = soup.find('dl')
plaintext = str(poem.get_text())
print(plaintext)
```
