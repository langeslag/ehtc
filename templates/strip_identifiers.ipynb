# Stripping Segment Identifiers

This code template demonstrates how to strip line-initial segment identifiers from a plaintext corpus and write it back to disk. We will use [ECHOE](https://github.com/ECHOEProject/echoe) as an example.

For more context and examples, see [`../demo/identifiers.ipynb`](https://github.com/langeslag/ehtc/blob/main/demo/identifiers.ipynb).

For this template we'll rely on the `pathlib` library for disk operations. This is one option among a number of eligible candidates, including `os` and `glob`, so take your pick:

```python
from pathlib import Path
```

You should already have ECHOE in `../corpora/echoe/`; if not, run the `fetch_repo.ipynb` template from the folder containing the present notebook (or from another subfolder one below repo root) first. We'll now identify our paths:

```python
echoe_path = '../corpora/echoe/plaintext/'
target_path = '../corpora/echoe-nlp/'
Path(target_path).mkdir(parents=True, exist_ok=True)
echoe_plain = Path(echoe_path)
```

Finally, we cycle through all plaintext files encountered in `echoe_path`, strip segment identifiers, and write them back to a new location `target_path`:

```python
for file in echoe_plain.glob('*txt'):
    outlines = []
    outfile = target_path + file.name
    inlines = open(file).read().splitlines()
    for line in inlines:
        if ': ' in line:
            line = line.split(': ', 1)[1]
        outlines.append(line)
    outdoc = '\n'.join(outlines)
    with open(outfile, 'w') as f:
        f.write(outdoc)
        
```
