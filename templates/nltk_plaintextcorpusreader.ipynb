# NLTK's `PlaintextCorpusReader()` Class

Just point `root` at your folder of plaintext files.

```python
from pathlib import Path
from nltk.corpus import PlaintextCorpusReader
```

```python
# Convert Path() to string or PlaintextCorpusReader() won't take it:
root = str(Path.cwd().parent / 'corpora' / 'echoe-bare')
echoe = PlaintextCorpusReader(root, '.*txt')
```

```python
echoe.words('394.11.txt')[:20]
```

```python
echoe.raw('394.11.txt')[:200]
```
